# 1. Boxplot (diagrama de caja)
🔹 Qué muestra

El boxplot resume la distribución de los datos usando estadísticos básicos:

Mediana (Q2) → la línea dentro de la caja.

Cuartiles (Q1 y Q3) → los bordes inferior y superior de la caja.

Rango intercuartílico (IQR) = Q3 − Q1.

Bigotes (whiskers) → normalmente se extienden hasta 1.5 × IQR o hasta el valor máximo/mínimo dentro de ese rango.

Outliers (valores atípicos) → puntos fuera de los bigotes.

🔹 Ventajas

Sencillo y muy útil para comparar medianas entre grupos.

Permite detectar asimetría y valores atípicos.

🔹 Desventajas

No muestra la forma completa de la distribución (por ejemplo, si hay varios picos o huecos).

# 2. Violinplot
🔹 Qué muestra

El violinplot combina un boxplot con una estimación de la densidad (parecido a una curva suavizada).
Es como un “violín” porque muestra la forma de la distribución reflejada a ambos lados.

🔹 Ventajas

Muestra tanto los estadísticos (como un boxplot) como la densidad completa.

Permite ver si la distribución es bimodal, asimétrica o aplanada.

Ideal para comparar formas de distribución entre grupos.

🔹 Desventajas

Un poco más difícil de interpretar para principiantes.

Puede engañar si el tamaño de muestra es muy pequeño (la densidad puede ser inestable).

----------------------------------------------------------------------------
## Estimación de densidad
La estimación de la densidad es una técnica estadística que se usa para describir cómo se distribuyen los datos continuos (por ejemplo, alturas, ingresos, temperaturas, etc.) a lo largo de un rango de valores.

En lugar de agrupar los datos en intervalos como hace un histograma, la estimación de densidad traza una curva suave que representa la probabilidad (o frecuencia relativa) de que los datos tomen determinados valores.

Imaginá que tenés muchos datos (por ejemplo, los pesos de 1000 personas).
Un histograma te mostraría barras con cuántas personas están en cada rango de peso (60–65 kg, 65–70 kg, etc.).
La estimación de densidad en cambio traza una curva continua que muestra dónde se concentran los valores y qué tan dispersos están.

## Tipos de estimación de densidad

 - Paramétrica

Supone que los datos siguen una distribución conocida (por ejemplo, normal o exponencial).

Solo se estiman los parámetros (media, desviación estándar, etc.).

 - No paramétrica (la más común)

No asume una forma específica para la distribución.

El método más usado es el Kernel Density Estimation (KDE).
---------------------------------------------------------------------

## Kernel Density Estimation (KDE)

La estimación de densidad por kernel suaviza los datos colocando una pequeña “campana” (kernel) sobre cada punto y sumando todas esas campanas para formar una curva continua.
El parámetro clave es el ancho de banda (bandwidth):

Si es demasiado pequeño, la curva será muy irregular (sobreajuste).

Si es demasiado grande, la curva será demasiado suave y perderá detalle.

#3. Swarmplot
🔹 Qué muestra

El swarmplot (también llamado dot swarm plot) muestra cada punto individual de los datos, pero los coloca de forma que no se superpongan (como un enjambre).

🔹 Ventajas

Muestra todos los datos sin ocultarlos.

Perfecto para conjuntos pequeños o medianos donde interesa ver cada valor.

Se puede combinar con un boxplot o violinplot para contexto estadístico.

🔹 Desventajas

En muestras grandes, puede resultar ruidoso o poco claro.

No resume los datos (no muestra mediana o cuartiles).
