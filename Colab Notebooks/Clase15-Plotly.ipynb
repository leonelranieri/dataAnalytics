La visualización interactiva permite que los usuarios interactúen con los datos: explorar, examinar y manipular la información.
Plotly es una librería de Python, permite generar gráfico es línea, facilitando su comparación, como fuera de línea brindando flexibilidad en su uso.


# Gráfico de dispersión - Scatter plot

```python
import plotly.express as px
import pandas as pd
```

```python
data = pd.DataFrame({
    'variable x' : [1,2,3,4,5],
    'variable y' : [2,4,1,3,5]
})
data
```

```python
fig = px.scatter(data,
                 x='variable x',
                 y='variable y',
                 title='Relación entre x e y')
fig.show()
```

###Personalización:
modificar colores, tamaños de puntos y añadir etiquetas.

```python
fig = px.scatter(data,
                 x='variable x',
                 y='variable y',
                 color='variable y',
                 size='variable x',
                 title='Relación entre x e y')
fig.update_layout(title='Relación en tre x e y (personalizado)')
fig.show()
```

# Gráfico de barras - Bar chart

```python
data1 = pd.DataFrame({
    'Categoría' : ['A','B','C','D'],
    'Valor' : [10, 15, 7, 12]
})
data1
```

```python
fig1 = px.bar(data1,
              x='Categoría',
              y='Valor',
              title='Comparación de categorías')
fig1.show()
```

### Personalizar:
ajustar colores, añadir etiquetas y horientar las barras horizontalmente

```python
fig1 = px.bar(data1,
              x='Categoría',
              y='Valor',
              color='Categoría',
              hover_data=['Valor'], #personaliza el tooltip
              hover_name='Categoría', #personaliza el nombre del punto
              labels={'Valor':'Importancia',
                      'Categoría':'Tipo'},
              orientation='v')
fig1.update_layout(title='Comparación de categorías (personalizado)')
fig1.show()
#
```

# Histograma

```python
import numpy as np

data2 = pd.DataFrame({
    'Valores': np.random.randn(100)
})
data2
```

```python
fig2 = px.histogram(data2,
                    x='Valores',
                    title='Distribución de valores')
fig2.show()
```

### Personalización:
definir el número de bins, cambiar colores y superponer una curva de densidad

```python
fig2 = px.histogram(data2,
                    x='Valores',
                    nbins=30,
                    color_discrete_sequence=['#EB89B5'])
fig2.update_layout(title='Distribución de valores (personalizado)')
fig2.show()
```

# Gráfico de líneas - Line Chart

```python
data3 = pd.DataFrame({
    'tiempo': range(10),
    'valor' : [2, 4, 1, 3, 5, 7, 6, 8, 9, 10]
})
data3
```

```python
fig3 = px.line(data3,
               x='tiempo',
               y='valor',
               title='Evolución en el tiempo')
fig3.show()
```

### Personalizar:
modificar el color de la línea, añadir marcadores, y mostrar rangos de confianza

```python
fig3 = px.line(data3,
               x='tiempo',
               y='valor',
               title='Evolución en el tiempo (personalizado)',
               line_shape='spline', # suaviza las curvas
               render_mode='svg')
fig3.show()
```

#Dashboard :
un dashboar integra múltioples gráficosy controles interactivos en un mismo espacio. Plotly junto con 'ipywidgets' simplifican el proceso de creación de un dashboar.

```python
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display
```

```python
# Datos de ejemplo
df = pd.DataFrame({
    'Categoria' : ['A','B','C','D'],
    'Valor_2023' : [10, 15, 7, 12],
    'Valor_2024' : [14, 18, 9, 16]
})
df
```

```python
# Dropdown para seleccionar el año
selector = widgets.Dropdown(
    options=['Valor_2023', 'Valor_2024'],
    value='Valor_2023',
    description='Año:',
    style={'description_width': 'initial'}
)
```

```python
# Función de actualización
def mostrar_grafico(columna_seleccionada):
    fig = px.bar(df,
                 x='Categoria',
                 y=columna_seleccionada,
                 title=f'Comparación de categorías - {columna_seleccionada[-4:]}',
                 labels={columna_seleccionada: 'Valor'})
    fig.show()
```

```python
# Ejecutar por primera vez
mostrar_grafico(selector.value)
```

```python
# Crear widget interactivo
interaccion = widgets.interactive(mostrar_grafico, columna_seleccionada=selector)

# Mostrar widget y gráfico interactivo
display(interaccion)
```
