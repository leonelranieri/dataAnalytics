#Consolidación de datos (ETL : extraer-transformar-cargar)
Es el proceso de combinar datos dispersos,a menudo proveniente de diferentes sistemas o formatos, en una única representación coherente. Permite obtener una visión más clara de la información y facilita la identificación de patrones, tendencias y relaciones dentro de datos.
##Pasos para la consolidicación de datos
1. Importar bibliotecas.
2. Cargar el conjunto de datos.
3. Exploración inicial: head(), info(), describe().
4. Limpieza de datos: eliminación de duplicados, manejo de valores faltantes y correción de formatos incosistentes.
5. Transformación de datos: creación de nuevas columnas, conversión de tipos de datos y normalización de valores.
6. Consolidación de datos: combinar múltiples DataFrames. Merge, concat y join.
7. Agrupación y agregación: la agrupación de datos es útil para realizar análisis descriptivos. Método groupby para agrupar según ciertas características y aplicar funciones de agregación.
