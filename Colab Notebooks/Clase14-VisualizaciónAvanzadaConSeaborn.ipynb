#Gráfico de líneas
Para mostrar la evolución de una variable durante un período de tiempo.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.DataFrame({
    'Tiempo' : [1, 2, 3, 4, 5],
    'Ventas' : [10, 15, 20, 25, 30]
 })
```

```python
from matplotlib import colors
sns.lineplot(x='Tiempo',
             y='Ventas',
             data=data,
             marker='o',
             color='green',
             linewidth=2)
plt.title('Líneas - Evolución de Ventas a lo largo del Tiempo')
plt.xlabel('Tiempo')
plt.ylabel('Ventas')
plt.xticks(data['Tiempo'])
plt.grid(True)
plt.show()
```

#Gráfico de barras
para comparar distintos grupos de datos

```python
categorias = ['A', 'B', 'C', 'D']
valores = [5, 7, 3, 8]

data_barras = pd.DataFrame({
    'Categoría': categorias,
    'Valor': valores
})

sns.barplot(x='Categoría',
            y='Valor',
            data=data_barras,
            palette='viridis'
            )

plt.title('Barras - Comparación de categorías')
plt.xlabel('Categoría')
plt.ylabel('Valor')
plt.show()
```

# Gráfico de dispersión
permite visualizar relación entre variables

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 1, 4]

data_dispersion = pd.DataFrame({
    'Variable x': x,
    'Variable y': y
})
```

```python
sns.scatterplot(x='Variable x',
                y='Variable y',
                data=data_dispersion,
                hue='Variable y',
                palette='viridis'
                )

plt.title('Dispersión - Relación entre Variables x e y')
plt.xlabel('Variable x')
plt.ylabel('Variable y')
plt.xticks(data_dispersion['Variable x'])
plt.yticks(data_dispersion['Variable y'])
plt.show()
```

#Boxplot
refleja la distribución de los datos y valores atípicos

```python
data_box = pd.DataFrame({
    'Categoría': ['A', 'A', 'B', 'B', 'C', 'C'],
    'Valor': [1, 2, 5, 6, 2, 10]
})

sns.boxplot(x='Categoría',
            y='Valor',
            data=data_box,
            palette='Set2'
            )

plt.title('Boxplot - Distribución de Datos por Categoría')
plt.show()
```

#Violinplot
combina un boxplot con la densidad de la probabilidad de la variable

```python
sns.violinplot(x='Categoría',
               y='Valor',
               data=data_box,
               palette='muted',
               alpha=0.6
               )

plt.title('Violinplot - Distribución de Datos por Categoría')
plt.show()
```

# Swarm Plot
Son útiles para visualizar la distribución de los datos sobre una categoría y evitar la sobreposición de puntos.

```python
sns.swarmplot(x='Categoría',
              y='Valor',
              data=data_box,
              color='blue',
              size=8
              )

plt.title('Swarm Plot - Distribución de Datos por Categoría')
plt.show()
```

#Subplot
División de la figura principal en una cuadrícula de subgráficos.
Los subplots se pueden crear ajustando la disposición de filas y columnas para acomodar los diferentes gráficos.

##Ejemplo 1 : subplots básicos

```python
import matplotlib.pyplot as plt
import numpy as np

# Creación de datos
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Creación de la figura y subplots
fig, axs = plt.subplots(2, 1) # 2 filas y una columna

# Primer subplot : seno
axs[0].plot(x, y1, color='blue')
axs[0].set_title('Seno')
axs[0].set_xlabel('x')
axs[0].set_ylabel('sin(x)')

# Segundo subplot : coseno
axs[1].plot(x, y2, color='red')
axs[1].set_title('Coseno')
axs[1].set_xlabel('x')
axs[1].set_ylabel('cos(x)')

# Ajustar el layut
plt.tight_layout()

# Mostrar la figura
plt.show()
```

# Ejemplo 2 : subplots en una cuadrícula

```python
y3 = np.tan(x)

# Crear una figura y variso subplots en una cuadrícula 2x2
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Primer subplot : seno
axs[0, 0].plot(x, y1, color='blue')
axs[0, 0].set_title('Seno')
axs[0, 0].set_xlabel('x')
axs[0, 0].set_ylabel('sin(x)')

# Segundo subplot : coseno
axs[0, 1].plot(x, y2, color='red')
axs[0, 1].set_title('Coseno')
axs[0, 1].set_xlabel('x')
axs[0, 1].set_ylabel('cos(x)')

# Tecer subplot : tangente
axs[1, 0].plot(x, y3, color='green')
axs[1, 0].set_ylim(-10, 10)
axs[1, 0].set_title('Tangente')
axs[1, 0].set_xlabel('x')
axs[1, 0].set_ylabel('tan(x)')

# Gráfico vacío para el último cuadrante
axs[1, 1].axis('off')

plt.tight_layout()
plt.show()

# Ajustar el layout
plt.tight_layout()

# Mostrar la figura
plt.show()

```

# Ejemplo 3 : subplots con seaborn

```python
import seaborn as sns
import pandas as pd

# Cargar el conjunto de datos
tips = sns.load_dataset("tips")

# Crear una figura y variso subplots en una cuadrícula 2x2
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Gráfico de cajas de las propinas por sexo
sns.boxplot(x='sex', y='tip', data=tips, ax=axs[0, 0])
axs[0, 0].set_title('Propinas por sexo')

# Gráfico de dispersión de total bill vs tip
sns.scatterplot(x='total_bill', y='tip', data=tips, ax=axs[0, 1], hue='time')
axs[0, 1].set_title('diespersión del Total bill vs tip')

# Histograma de propinas
sns.histplot(tips['tip'], bins=20, ax=axs[1, 0], kde=True)
axs[1, 0].set_title('Histograma de distribución de propinas')

# Gráfico vacío
axs[1, 1].axis('off')

# Ajustar el layout
plt.tight_layout()

# Mostrar la figura
plt.show()
```
