# Tipos de data-viz básicos y sus personalizaciones





## Gráfico de líneas
Ideales para mostrar la evolución de una variable a lo largo del tiempo. Especialmente útiles en series temporales, donde queremos observar tendencias.

```python
import matplotlib.pyplot as plt

tiempo = [1, 2, 3, 4, 5]
ventas = [10, 15, 20, 25, 30]
```

```python
plt.plot(tiempo, ventas)
```

###Personalización del gráfico de líneas
- marker='s'

  Indica el símbolo usado para marcar los puntos de datos.
  's' significa square (cuadrado).
  Otros ejemplos:
  'o' → círculo
  '^' → triángulo hacia arriba
  'x' → cruz
  'D' → rombo
  '*' → estrella
  📊 Cada punto (x, y) en la gráfica se mostrará con ese símbolo.

- linestyle='--'
  Controla el estilo de la línea que une los puntos.
  '--' significa línea discontinua (guiones).
  Otros estilos posibles:
  '-' → línea continua
  ':' → línea punteada
  '-.' → guiones y puntos alternados
  '' o 'None' → sin línea (solo marcadores)

- plt.grid(True)
  Activa una cuadrícula de fondo (líneas horizontales y verticales).
  Ayuda a leer mejor los valores.

- plt.xticks(tiempo)
  Controla las marcas del eje X (los valores donde se colocan las etiquetas).
  En este caso, estás diciendo que se usen exactamente los valores de la lista tiempo como posiciones para las marcas del eje X.

```python
plt.plot(tiempo, ventas, marker='s', linestyle='--', color='green', linewidth=2, markersize=10)
plt.xlabel('Tiempo', fontsize=10)
plt.ylabel('Ventas', fontsize=10)
plt.title('Evolución de Ventas a lo Largo del Tiempo', fontsize=14)
plt.grid(True)
plt.xticks(tiempo)
plt.show()
```

## Gráfico de barras

Excelentes para compararar diferentes tipos de datos

```python
categorias = ['A', 'B', 'C', 'D']
valores = [5, 7, 3, 8]
```

```python
plt.bar(categorias, valores)
plt.show()
```

### Personalización del gráfico de barras   


```python
plt.bar(categorias,
  valores,
  color=['red', 'blue', 'green', 'purple'],
  width=0.5)

plt.xlabel('Categorías', fontsize=12)
plt.ylabel('Valores', fontsize=12)
plt.title('Comparación de categorías', fontsize=14)
plt.xticks(rotation=45)
plt.show()
```

## Gráfico de dipersión

Permiten visualizar la relación entre dos variables.
ütil para identificar correlaciones.

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 1 ,4]

plt.scatter(x, y)
plt.show()
```

Personalización del gráfico de dispersión

- s:
  establece el tamaño de los puntos en el gráfico

```python
plt.scatter(x, y, color='cyan', s=[100, 150, 200, 250, 300], marker='o')
plt.xlabel('Variable X', fontsize=12)
plt.ylabel('Variable Y', fontsize=12)
plt.title('Gráfico de Dispersión', fontsize=14)
plt.grid(True)
plt.show()
```
