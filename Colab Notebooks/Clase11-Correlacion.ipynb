# Correlación
Es una medida estadística que describe la relación entre dos variables. Es decir, que un cambio en una variable tiende a estar acompañado de un cambio en la otra.
Analizar la correlación entre variables permite identificar patrones y tendencias en los datos.
## Tipos de correlación
- Correlación positiva: cuando una variable aumenta la otra también.
- Correlación negativa: cuando una variable aumenta la otra disminuye.
- Correlación nula: no hay una relación discernible entre las variables.
# Medidas de correlación.
Para cuantificar la correlación se utiliza el coeficiente de correlación de Pearson, que mide la fuerza y la dirección de la relación lineal entre dos variables continuas.
Que dos variables tengan una alta correlación, no implica que el cambio del valor de una de las variables, sea la causa del cambio en la otra. Solo indica que ambas variables cambian en la misma proporción.
## Correlación no implica causalidad.

```python
# Ejemplo práctico

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df_iris = sns.load_dataset('iris')
```

```python
df_iris['species'].unique()
```

```python
setosa = df_iris[df_iris['species'] == 'setosa']
# obtener solo variables numéricas
setosa.drop(['species'], axis='columns', inplace=True)
setosa
```

```python
# crear matriz de correlación
correlation_matrix = setosa.corr()
correlation_matrix
```

```python
# es posible explorar el valor de correlación entre dos variables
float(correlation_matrix['sepal_length']['sepal_width'])
```

```python
# visualmente la mejor opción es el gráfico de dispersión.
sns.scatterplot(x='sepal_length', y='sepal_width', data=setosa)
plt.show()
```

```python
# exploración visual del mapa de correlaciones con el mapa de calor.

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.show()
```
