# Actividad 1: Análisis de Ventas a lo Largo del Tiempo

Crear un gráfico de líneas que muestra las ventas mensuales.

Redactar un párrafo breve que explique la gráfica.

```python
import matplotlib.pyplot as plt
```

```python
# Datos de ventas mensuales (en unidades)
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'] # Lista de los meses del año
ventas = [100, 120, 130, 150, 180, 170, 200, 210, 250, 270, 300, 320] # Lista de las unidades vendidas correspondientes a cada mes
```

```python
from matplotlib import lines
plt.plot(meses, ventas, marker='o', linestyle='-.', color='g', markersize=7, linewidth=2)
plt.title('Ventas Mensuales', fontsize=14)
plt.xlabel('Mes', fontsize=12)
plt.ylabel('Unidades Vendidas', fontsize=12)
plt.grid(True)
plt.show()
```

En la primera mitad del año las ventas crecen significativamente, luego se produce una baja entre Mayo y Junio para repuntar nuevamente hasta fin de año con un crecimiento exponencial.

# Actividad 2: Comparación de Productos en una Campaña
Consigna:

Crear un gráfico que muestre las ventas por artículo.

Agregar el contexto al gráfico y buscar una forma de resaltar el artículo más vendido

```python
# Datos de ventas por producto
productos = ['Producto A', 'Producto B', 'Producto C', 'Producto D']  # Lista de los nombres de los productos
ventas = [300, 250, 400, 350] # Lista de las unidades vendidas de cada producto en la campaña.
```

```python
plt.bar(productos,
        ventas,
        color=['lightblue', 'lightblue', 'purple', 'lightblue'],
        width=0.5)
plt.title('Ventas por Producto', fontsize=14)
plt.xlabel('Producto', fontsize=12)
plt.ylabel('Unidades Vendidas', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

De los productos comparados se destaca el 'Producto C' como el más vendido.
