```python
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from ipywidgets import interact
from IPython.display import display

data = pd.read_csv('https://drive.google.com/uc?id=1AmoD9mPsv87mcY9p5-03G18KkxqmuGzP')
```

###Funciones

```python
def obtenerMes(fecha):
    return fecha.month

def cambiarFormatoFecha(fecha):
    return fecha.strftime('%d/%m/%Y')

def aplicarMayuscula(cadena):
    return cadena.capitalize()
```

### Cambios en formato de fecha y ventas

```python
data['fecha'] = pd.to_datetime(data['fecha'])

data['ventas'] = data['ventas'].astype(str)
data['ventas'] = data['ventas'].str.replace('$', '')
data['ventas'] = data['ventas'].astype(float)

data['region'] = data['region'].astype(str)
data['region'] = data['region'].apply(aplicarMayuscula)
```

## Primer gráfico :
de líneas, muestra la evolución en el tiempo de las ventas de los productos

```python
data['mes'] = data['fecha'].apply(obtenerMes)

ventas_mensuales_por_producto = data.groupby(['mes', 'producto'])['ventas'].sum().reset_index()

fig_line = px.line(ventas_mensuales_por_producto,
              x='mes',
              y='ventas',
              color='producto',
              markers=True,
              line_shape='spline',
              labels={'producto': 'Producto', 'mes': 'Mes', 'ventas': 'Total Vendido($)'})

fig_line.update_layout(title={
        "text": "Evolución mensual de ventas",
        "x": 0.5,           # centrar el título
        "xanchor": "center",
        "font": {"size": 24}},
        xaxis_title="Mes",
        yaxis_title="Ventas")

fig_line.update_xaxes(
    dtick=1,
)  # permite mostrar todos los meses, sino aparecen de dos en
   # dos debido al autotick para no sobrecargar el eje

fig_line.show()
```

###Conclusiones:
El smart tv alcanzó el máximo de ventas en el mes de Noviembre entre todos lo productos. El primer pico fue en Septiembre luego de su peor caída en ventas en Agosto.
Los auriculares alcanzaron su caída en ventas en febrero para luego alcanzar su primer pico en ventas al mes siguiente. Y entre Agosto y Septiembre logro ser el segundo producto mmás vendido.
Las notebooks se mantuvieron estable en ventas durangte todo el año, rondando el promedio total.
El smarthphone obtuvo pico en ventas en Octubre y la consola de videojuegos mantuvo sus ventas dentro del promedio del total de todos los productos.

## Segundo Gráfico :
de barras, para comparar las ventas totales de los productos

```python
ventas_totales_productos = data.groupby('producto')['ventas'].sum().reset_index()

fig_bar = px.bar(ventas_totales_productos,
             x='producto',
             y='ventas',
             color='producto',
             labels={'ventas': 'Total Vendido($)'},
             hover_name='producto',
             hover_data={'producto': False})

fig_bar.update_layout(title={
        "text": "Ventas totales por producto",
        "x": 0.5,           # centrar el título
        "xanchor": "center",
        "font": {"size": 24}})

fig_bar.show()
```

### Conclusión: las ventas totales de los productos no presentan grandes diferencias, el productos más vendido son los auriculares y erl menos vendido la consola de videjuegos.

##Tercer Gráfico:
visualizar ventas de una región mediante filtros

```python
ventas_de_productos_por_region = data.groupby(['region', 'producto'])['ventas'].sum().reset_index()

#Dropdown para seleccionar la región
selector = widgets.Dropdown(
    options=ventas_de_productos_por_region['region'].unique(),
    value=ventas_de_productos_por_region['region'].unique()[0],
    description='Región:',
)

#Función de actualización
def mostrar_grafico(region):
    datos_filtrados = ventas_de_productos_por_region[ventas_de_productos_por_region['region'] == region]
    fig = px.bar(datos_filtrados,
                 x='producto',
                 y='ventas',
                 color='producto',
                 labels={'ventas': 'Total Vendido($)'},
                 hover_name='producto',
                 hover_data={'producto': False},
                 title=f'Ventas por producto en la región {region}')


    fig.show()

mostrar_grafico(selector.value)

widgets.interactive(mostrar_grafico, region=selector)
```

## Conclusión:
- Zona Centro : se visualiza la consola como el más vendido, y tanto el smart como smartphone son los productos menos vendidos.
- Zona Norte : notebook resalta como el productos más vendido, y la consola como el menos vendido.
- Zona Sur : predominan las ventas de los auriculares y la notebbok se encansilla en los menos vendidos.
- Zona Oeste : el smartphone es el producto más vendido y los auriculares los que menos se vendieron.
- Zona Este : el producto más vendido es el smart y el menos vendido la consola.

```python
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from ipywidgets import interact
from IPython.display import display

data = pd.read_csv('https://drive.google.com/uc?id=1wALe5NDEeTXLusrQRbPgd8sQVjOSJww8')
data
```

```python
data['fecha'] = pd.to_datetime(data['fecha'])
data = data.sort_values(by='fecha')
```

```python
seguidores_enElTiempo_porPlataforma = data.groupby(['fecha', 'plataforma'])['seguidores'].sum().reset_index()
```

```python
selector = widgets.Dropdown(
    options=seguidores_enElTiempo_porPlataforma['plataforma'].unique(),
    value=seguidores_enElTiempo_porPlataforma['plataforma'].unique()[0],
    description='Plataforma:',
)

def mostrar_grafico(plataforma):
    datos_filtrados = seguidores_enElTiempo_porPlataforma[seguidores_enElTiempo_porPlataforma['plataforma'] == plataforma]
    fig = px.line(datos_filtrados,
                 x='fecha',
                 y='seguidores',
                 title=f'Seguidores por fecha en {plataforma}',
                 line_shape='spline',
                 render_mode='svg')
    fig.show()

mostrar_grafico(selector.value)

widgets.interactive(mostrar_grafico, plataforma=selector)
```

```python
meGusta_y_comentarios_porPlataforma = data.groupby(['plataforma'])[['likes', 'comentarios']].sum().reset_index()
meGusta_y_comentarios_porPlataforma
```

```python
selector = widgets.Dropdown(
    options=meGusta_y_comentarios_porPlataforma['plataforma'].unique(),
    value=meGusta_y_comentarios_porPlataforma['plataforma'].unique()[0],
    description='Plataforma:',
)

def mostrar_grafico(plataforma):
    fila = meGusta_y_comentarios_porPlataforma[meGusta_y_comentarios_porPlataforma['plataforma'] == plataforma].iloc[0]

    df_fila = pd.DataFrame({'categorias': ['likes', 'comentarios'],
                            'valores': [fila['likes'], fila['comentarios']]})

    fig = px.bar(df_fila,
                 x='categorias',
                 y='valores',
                 color='categorias',
                 title=f'Comparación de likes y comentarios en {plataforma}',
                 labels={'categorias': 'Interacción', 'valores': 'Cantidad'},
                 )
    fig.show()

mostrar_grafico(selector.value)

widgets.interactive(mostrar_grafico, plataforma=selector)
```

```python
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display

# Aseguramos que la fecha sea datetime
data['fecha'] = pd.to_datetime(data['fecha'])

# Lista de plataformas disponibles
plataformas = data['plataforma'].unique().tolist()

# ---- WIDGETS ----

# 1) Selector de varias plataformas
selector_plataformas = widgets.SelectMultiple(
    options=plataformas,
    description='Plataformas:',
    value=[plataformas[0]],
    rows=6
)

# 2) Selector de fecha inicial y final
selector_fecha_inicio = widgets.DatePicker(
    description='Fecha inicio:',
    value=data['fecha'].min().date()
)

selector_fecha_fin = widgets.DatePicker(
    description='Fecha fin:',
    value=data['fecha'].max().date()
)

# ---- FUNCIÓN PRINCIPAL ----
def mostrar_grafico(plataformas_sel, fecha_ini, fecha_fin):

    if not plataformas_sel:
        print("Seleccioná al menos una plataforma.")
        return

    # Convertimos fechas a datetime
    fecha_ini = pd.to_datetime(fecha_ini)
    fecha_fin = pd.to_datetime(fecha_fin)

    # 1) Filtrar plataformas
    df_filtrado = data[data['plataforma'].isin(plataformas_sel)]

    # 2) Filtrar por rango de fechas
    df_filtrado = df_filtrado[
        (df_filtrado['fecha'] >= fecha_ini) &
        (df_filtrado['fecha'] <= fecha_fin)
    ]

    if df_filtrado.empty:
        print("No hay datos para los filtros seleccionados.")
        return

    # 3) Agrupamos
    df_agg = df_filtrado.groupby("plataforma")[["likes", "comentarios"]].sum().reset_index()

    # 4) Convertimos a formato largo para Plotly
    df_melt = df_agg.melt(
        id_vars="plataforma",
        value_vars=["likes", "comentarios"],
        var_name="tipo",
        value_name="cantidad"
    )

    # 5) Graficamos
    fig = px.bar(
        df_melt,
        x="plataforma",
        y="cantidad",
        color="tipo",
        barmode="group",
        title=f"Actividad por plataforma ({fecha_ini.date()} → {fecha_fin.date()})",
        text="cantidad"
    )
    fig.update_traces(textposition='outside')
    fig.show()

mostrar_grafico(selector_plataformas.value,
                selector_fecha_inicio.value,
                selector_fecha_fin.value)

# ---- CONECTAR WIDGETS ----
widgets.interact(
    mostrar_grafico,
    plataformas_sel=selector_plataformas,
    fecha_ini=selector_fecha_inicio,
    fecha_fin=selector_fecha_fin
)
```
